4. mutable, size- immutable

```python
s1 = pd.Series([10,20,30,40,3.14])
s1 = pd.Series(['a','b','c'])
s1= pd.Series([10,20,30,40,'a'])
s1
```

```python
s1 = pd.Series([10,20,30,40], index = ['a','b','c','d'])
s1
```

```python
s2 = pd.Series([50,60,70,80], index = ['c','d','e','f'])
s2
```
